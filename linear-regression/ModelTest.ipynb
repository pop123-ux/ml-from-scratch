```python
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.datasets import make_regression
import matplotlib.pyplot as plt
from LinearRegression import LinearRegression
```

```python
n_samples = 1000
n_features = 10
```

```python
X, y = make_regression(n_samples=n_samples, n_features=n_features, n_informative=5, noise=10, random_state=42)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=1)
```

```python
X
```

```python
fig = plt.figure(figsize=(6,4))
plt.scatter(X[:, 0], y, c='b', marker='o', s=5)
plt.show()
```

```python
def mse(y, predictions):
    return np.mean(np.square(y - predictions))
```

```python
model = LinearRegression(lr=0.007)
model.fit(X_train, y_train)
train_predictions = model.predict(X_test)
```

```python
mse(y_test, train_predictions)
```

```python
y_pred_train = model.predict(X_train)

print(mse(y_train, y_pred_train))
fig = plt.figure(figsize=(6,4))
plt.scatter(y_train, y_pred_train, c='g', s=10)
plt.plot([min(y) , max(y)], [min(y), max(y)], c='r', linewidth=2)
plt.xlabel('True values')
plt.ylabel('Predicted Values')
plt.show()
```

```python
y_pred_test = model.predict(X_test)

print(mse(y_pred_test, y_test))

fig = plt.figure(figsize=(6,4))
plt.scatter(y_test, y_pred_test, c='g', s=10)
plt.plot([min(y), max(y)], [min(y), max(y)], c='r', linewidth=2)
plt.xlabel('True values')
plt.ylabel('Predicted values')
plt.show()
```

```python
# Now we are going to try L2 regularization
model = LinearRegression(lr=0.01, penalty='l2')
model.fit(X_train, y_train)
train_predictions = model.predict(X_train)
```

```python
mse(y_train, train_predictions) # It looks like it's a bit overfitting
```

```python
y_pred_train = model.predict(X_train)

fig = plt.figure(figsize=(6,4))
plt.scatter(y_train, y_pred_train, c='g', s=10)
plt.plot([min(y), max(y)], [min(y), max(y)], c='r', linewidth=2)
plt.xlabel('True values')
plt.ylabel('Predicted values')
plt.show()
```

```python
y_pred_test = model.predict(X_test)

print(mse(y_test, y_pred_test))
fig = plt.figure(figsize=(6,4))
plt.scatter(y_test, y_pred_test, c='g', s=10)
plt.plot([min(y), max(y)], [min(y), max(y)], c='r', linewidth=2)
plt.xlabel("True values")
plt.ylabel("Predicted values")
plt.show()
```
